```python
# Accounts Without Transactions
number_of_accounts = accounts_df['account_id'].count()
number_accounts_with_transation = len(transactions_df['account_id'].unique().tolist())
print(
    f'The number of accounts in the dataset is {number_of_accounts}',
    f'\nThe number of accounts with at least one transaction is {number_accounts_with_transation}'
)
```

```python
# Primary Key Uniqueness Validation
primary_keys = [accounts_df['account_id'], customers_df['customer_id'], transactions_df['transaction_id']]


for pk in primary_keys:
    if pk.duplicated().any():
        print(f'Duplicate values were found in the {pk.name} column.')
        continue
    else:
        print(f'The {pk.name} column does not contain duplicate values.')
```

```python
# Regional Data Consistency Check

region_without_customers = []

for region in regions_regions:
    if region not in customers_regions:
        region_without_customers.append(region)
        continue

if len(region_without_customers) == 0:
    print('At least one customer is assigned to each region.')
else:
    print(f'The following regions have no assigned customers: {region_without_customers}')

missing_regions = []

for region in customers_regions:
    if region not in regions_regions:
        missing_regions.append(region)
        continue

if len(missing_regions) == 0:
    print('No customer is assigned to a region that does not exist in the regions table.')
else:
    print(f'The customers table contains regions that are not present in the regions table: {missing_regions}')
```

```python
# Validation of Primary and Foreign Key Set Consistency Between Tables

if set(customers_df['customer_id'].unique()) == set(accounts_df['customer_id'].unique()):
    print('The primary key in the customers table and the foreign key in the accounts table are consistent.')
else:
    print('WARNING! The primary key in the customers table and the foreign key in the accounts table are not consistent.')
```

```python
# Validation of Accounts Without Transactions and Their Statuses
accounts_accounts_id = accounts_df['account_id'].unique().tolist()
transactions_accounts_id = transactions_df['account_id'].unique().tolist()
account_without_transaction = []

for id in accounts_accounts_id:
    if id not in transactions_accounts_id:
        account_without_transaction.append(id)
        continue

account_without_transaction_df = None

if len(account_without_transaction) > 0:
    print(f'There are {len(account_without_transaction)} accounts without transactions.')
    account_without_transaction_df = accounts_df[accounts_df['account_id'].isin(account_without_transaction)]

account_without_transaction_df['status'].unique()
```

```python
# Account Counts Without Transactions by Status
account_without_transaction_df['status'].value_counts()
```

```python
# Distribution of Accounts by Opening Date
account_without_transaction_df.describe()
```

```python
# Analysis of Opening Dates for Active Accounts
# This analysis is important for assessing data quality. An active account without transactions for more than 180 days would normally be expected to have a dormant status.
account_without_transaction_df[account_without_transaction_df['status'] == 'active'].describe()
```

```python
# Distribution of Dormant Accounts by Opening Date
# This analysis examines whether dormant accounts are consistent with the business definition of dormancy. If an account has been open for less than 180 days and is already marked as dormant, it may indicate a potential data quality issue.
account_without_transaction_df[account_without_transaction_df['status'] == 'dormant'].describe()
```

```python
# Distribution of Accounts Without Transactions by Status Over Time
awt_hist = px.histogram(
    account_without_transaction_df,
    x='open_date',
    title='Distribution of Accounts Without Transactions by Status Over Time',
    color='status',
    barmode='group',
    
)

awt_hist.update_xaxes(title=None)
awt_hist.update_yaxes(title=None)
awt_hist.update_layout(
    legend=dict(
        orientation='h',
        yanchor='bottom',
        y= -0.2,
        xanchor='center',
        x=0.5,),
    title={
        'x': 0.5,
        'y': 0.85,
        'pad': {'t': 5},},
    autosize=True
)

awt_hist.show()
```

```python
# Distribution of Accounts Without Transactions by Account Type Over Time
# Verification of Whether the Active/Dormant Status Assignment Issue Is Limited to a Specific Account Type or Time Period
awt_hist_type = px.histogram(
    account_without_transaction_df,
    x='open_date',
    title='Distribution of Accounts Without Transactions by Account Type Over Time',
    color='account_type',
    barmode='group',
    
)

awt_hist_type.update_xaxes(title=None)
awt_hist_type.update_yaxes(title=None)
awt_hist_type.update_layout(
    legend=dict(
        orientation='h',
        yanchor='bottom',
        y= -0.2,
        xanchor='center',
        x=0.5,),
    title={
        'x': 0.5,
        'y': 0.85,
        'pad': {'t': 5},},
    autosize=True,
    legend_title_text='Account Type'
)

awt_hist_type.show()
```

```python
# Account Counts by Status and Account Type
awt_status_type = (
    account_without_transaction_df[['status','account_type']]
    .value_counts()
    .reset_index(name='count')
)

awt_status_type.rename(columns={'status': 'Status', 'account_type': 'Account Type'}, inplace=True)

awt_status_type
```

```python
# Distribution of Accounts by Status and Account Type

awt_status_type_plot = px.bar(
    awt_status_type,
    x="Status",
    y='count',
    color='Account Type',
    barmode='group',
    title='Distribution of Accounts by Status and Account Type'
)

awt_status_type_plot.update_layout(
    legend=dict(
            orientation='h',
            yanchor='bottom',
            y= -0.2,
            xanchor='center',
            x=0.5,),
        title={
            'x': 0.5,
            'y': 0.85,
            'pad': {'t': 5},},
    title_x = 0.5,
    legend_title_text='Account Type'
)
awt_status_type_plot.update_xaxes(title=None)
awt_status_type_plot.update_yaxes(title=None)

awt_status_type_plot.show()
```

```python
# Filtering Savings and Credit Accounts
saving_accounts_df = accounts_df[accounts_df['account_type']=='savings']
credit_accounts_df = accounts_df[accounts_df['account_type']=='credit']
```

```python
# Analysis of Transaction Types Associated with Savings Accounts
savings_transactions_df = transactions_df[transactions_df['account_id'].isin(saving_accounts_df['account_id'])]
savings_transactions_df['transaction_type'].unique()
```

```python
# Analysis of Transaction Types Associated with Credit Accounts
credits_transactions_df = transactions_df[transactions_df['account_id'].isin(credit_accounts_df['account_id'])]
credits_transactions_df['transaction_type'].unique()
```

```python
# Distribution of Absolute Transaction Amounts
abs_amount_transactions_df = transactions_df[['amount']].copy()
abs_amount_transactions_df['abs_amount'] = abs_amount_transactions_df['amount'].abs()

abs_amount_transactions_df.describe()
```

```python
# Visualization of the Distribution of Absolute Transaction Amounts
treshold = abs_amount_transactions_df['abs_amount'].quantile(0.95)
abs_amount_transactions_df['amount_bucket'] = abs_amount_transactions_df['abs_amount'].apply(
    lambda x: x if x <= treshold else 'others'
)

abs_amount_transactions_fig = px.histogram(
    abs_amount_transactions_df,
    x='amount_bucket',
    title='Distribution of Absolute Transaction Amounts',
)

abs_amount_transactions_fig.update_layout(
   title_x = 0.5,
)

abs_amount_transactions_fig.show()
```

## EDA Conclusions

The dataset passed all essential structural validation checks and was considered suitable for further analysis.

Several important limitations were identified:

- Account status labels are not fully consistent with observed transaction behaviour.
- A significant share of transactions occurred before the recorded account open date.
- The business meaning of customer value and customer activity required additional interpretation.
- Regional comparisons require normalization due to substantial differences in population size.

These findings directly influenced the analytical methodology used in the subsequent SQL analyses.
